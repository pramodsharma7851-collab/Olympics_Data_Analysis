```python
def fetch_medal_tally(df,year,country):
  medal_df=medal_tally.copy()
  flag=0
  if year=='Overall' and country=='Overall':
    temp_df=medal_df
  if year=='Overall' and country!='Overall':
    flag=1
  temp_df=medal_df[medal_df['region']==country]
  if year!='Overall' and country=='Overall':
    temp_df=medal_df[medal_df['Year']==int(year)]
  if year!='Overall' and country!='Overall':
    temp_df=medal_df[(medal_df['Year']==int(year)) & (medal_df['region']==country)]

  if flag==1:
      X=temp_df.groupby('Year').sum()[['Gold','Silver','Bronze']].sort_values(by='Year',ascending=True).reset_index()

  else:
      X = temp_df.groupby('region').sum()[['Gold','Silver','Bronze']].sort_values(by='Gold',ascending=False).reset_index()
  X['Total']=X['Gold'] +X['Silver'] +X['Bronze']

  print(X)

```

```python
def weight_v_height(df,sport, s=40, alpha=1):
    athlete_df=df.drop_duplicates(subset=['Name','region'])# Added .copy() to avoid SettingWithCopyWarning
    athlete_df['Medal'] = athlete_df['Medal'].fillna('No Medal') # Changed to explicit assignment to avoid FutureWarning
    temp_df=athlete_df[athlete_df['Sport']==sport]
    plt.figure(figsize=(14,14))
    # Removed: fig.update_xaxes(dtick=10) - This is for Plotly, not Matplotlib
    return sns.scatterplot(x=temp_df['Weight'],y=temp_df['Height'],hue=temp_df['Medal'],style=temp_df['Sex'], s=s, alpha=alpha)
```

```python
def wise(medal_tally,country1,country2):

        f=medal_tally[medal_tally['region']==country1].groupby(['region','Year'])[['Gold','Silver','Bronze']].sum().reset_index()
        f[country1]=f['Gold']+f['Silver']+f['Bronze']
        f=f[['Year',country1]]
        g=medal_tally[medal_tally['region']==country2].groupby(['region','Year'])[['Gold','Silver','Bronze']].sum().reset_index()
        g[country2]=g['Gold']+g['Silver']+g['Bronze']
        g=g[['Year',country2]]
        f1=f.merge(g,on='Year',how='left')
        fig=px.line(f1,x='Year',y=[country1,country2])
        fig=fig.update_layout(title='country1 vs country2',
                  xaxis_title='Year',
                  yaxis_title='Number of medals')
        return fig
```
